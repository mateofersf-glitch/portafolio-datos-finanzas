##Business_Understanding

###🏢 01. SaaS B2B FP&A

---

### 🌐 1. Modelo de Negocio & Contexto
* **Sector:** Software como Servicio (SaaS) B2B.
* **Modelo de Monetización:** Suscripciones recurrentes (planes mensuales y anuales) con la siguiente estructura financiera:
  * **Ingresos Recurrentes:** MRR (*Monthly Recurring Revenue*) y ARR (*Annual Recurring Revenue*).
  * **Costos de Servir (COGS):** Infraestructura en la nube (AWS/GCP), licencias de software de soporte, servidores y comisiones de pasarela de pago.
  * **Gastos Operativos (OpEx):**
    * **CAC (Costos de Ventas & Marketing):** Inversión directa para adquirir nuevos clientes.
    * **R&D (Investigación & Desarrollo):** Salarios del equipo de ingeniería de software y ciencia de datos.
    * **G&A (Gastos Generales & Administrativos):** Operaciones, legal y administración corporativa.

---

### ❓ 2. Preguntas de Negocio a Responder (Business Questions)
1. **Crecimiento & Recurrencia:** ¿Cómo evoluciona el ingreso recurrente mensual (MRR) desglosado en *Nuevo MRR*, *Expansion MRR* (upselling) y *Churn MRR* (cancelaciones)?
2. **Salud de Unit Economics:** ¿Es sostenible el modelo de adquisición? (¿El valor de vida del cliente $LTV$ supera al costo de adquisición $CAC$ por un ratio mínimo de 3:1?).
3. **Eficiencia y Eficiencia de Caja (Burn Rate):** ¿Cuántos meses de liquidez (*Runway*) quedan al ritmo actual de consumo de caja en relación a la facturación proyectada?
4. **Pronóstico Estratégico:** ¿Cuál es la proyección de MRR para los próximos 12 meses y qué nivel de *Churn* máximo puede tolerar la empresa antes de comprometer el flujo de caja operativo?

---

### 📊 3. Definición de KPIs Financieros y Analíticos de SaaS

Para responder a estas preguntas, construiremos y monitorearemos las siguientes métricas:

* **MRR (Monthly Recurring Revenue):**
  $$MRR_{t} = MRR_{t-1} + \text{Nuevo } MRR + \text{Expansión } MRR - \text{Churn } MRR$$

* **Churn Rate (Tasa de Cancelación de Clientes):**
  $$\text{Churn Rate} = \frac{\text{Clientes Perdidos en el Período}}{\text{Clientes Totales al Inicio del Período}} \times 100$$

* **LTV (Lifetime Value del Cliente):**
  $$LTV = \frac{\text{ARPU (Ingreso Promedio por Usuario)} \times \text{Margen Bruto \%}}{\text{Churn Rate}}$$

* **CAC Ratio & Payback Period:** Meses necesarios para recuperar el costo de adquirir un cliente con su margen bruto:
  $$\text{Payback Period (meses)} = \frac{CAC}{\text{ARPU} \times \text{Margen Bruto \%}}$$

* **Runway (Meses de Vida de Caja):**
  $$\text{Runway} = \frac{\text{Caja Disponible}}{\text{Burn Rate Neto Mensual}}$$

##Data_Collection

###📥 Ingesta de Datos Reales SaaS (Benchmark Financiero)

---

### 🎯 Objetivo
Extraer, consolidar y estructurar estados financieros reales desde la Bolsa de Valores de Nueva York (NYSE/NASDAQ) para 4 empresas líderes del sector SaaS B2B, permitiendo realizar un análisis comparativo (*Benchmark*) de ingresos, costos y márgenes operativos.

### 🐥 Explicación sencilla
En lugar de inventar números o usar tablas falsas, nos conectamos en tiempo real a la bolsa de valores para traer los balances reales de 4 empresas gigantes del software. Así podremos comparar cuál gana más dinero y cuál gasta mejor su presupuesto.

### 🏢 Las 4 Empresas SaaS Seleccionadas

   * **¿Qué significa el ticker?:** Proviene de la palabra en inglés **Shop** (Tienda).

---

### 🛠️ Estructura del Script de Extracción en Python
* **Librería principal:** `yfinance` (para la conexión a la API financiera) y `pandas` (para la manipulación de dataframes).
* **Entregable:** Matriz comparativa de Ingresos Totales (*Total Revenue*) y Costos Operativos (*OpEx*) expresada en dólares reales (USD).

```python
import yfinance as yf
import pandas as pd

# 1. Definimos nuestro grupo de 4 empresas SaaS
lista_saas = ["CRM", "SHOP", "TEAM", "NOW"]
datos_empresas = {}

print("⏳ Descargando estados financieros reales desde la Bolsa de Valores...")

# 2. Bucle (loop) para traer la data bruta de cada ticker
for ticker in lista_saas:
    empresa = yf.Ticker(ticker)
    datos_empresas[ticker] = empresa.financials

print("✅ ¡Datos brutos descargados exitosamente desde Yahoo Finance!\n")
```

##Data_Preparation

### 🧹Limpieza y Formateo Financiero

---

### 🎯 Objetivo
Transformar los datos financieros brutos extraídos de la API de Yahoo Finance en un DataFrame estructurado, normalizado a miles de millones de dólares ($B) y alineado por año calendario para su correcto análisis en la fase de EDA.

---

### 🐥 Explicación sencilla: ¿Por qué hay celdas N/A y años diferentes?

Cuando trabajamos con datos financieros reales de distintas empresas de Wall Street, nos encontramos con dos realidades fundamentales:

1. **La limitación de la API Gratuita (Ventana de 4 años):**
   * La librería `yfinance` en su versión pública solo nos entrega los **últimos 4 reportes anuales cerrados**. Por esa razón, los años anteriores a 2022 aparecen como `N/A` y fueron filtrados para mantener la tabla limpia.

2. **La diferencia de Años Fiscales (Fiscal Year vs. Calendar Year):**
   * No todas las empresas cierran sus libros contables el 31 de diciembre.
   * **`SHOP`**, **`TEAM`** y **`NOW`** usan el año calendario estándar (cierran en diciembre).
   * **`CRM` (Salesforce)** tiene un año fiscal desplazado: su año contable cierra a finales de enero. Por ello, su reporte del 31 de enero de 2026 corresponde en la práctica a los ingresos del año operativo previo, generando ese dato del **2026** que las demás empresas aún no reportan.

---

### 🛠️ Transformaciones Realizadas
* **Normalización de Escala:** Conversión de valores nominales a Billones de USD ($B$) dividiendo entre $1,000,000,000$.
* **Estandarización de Fechas:** Extracción exclusiva del año (YYYY) del índice temporal.
* **Depuración de Filas Incompletas:** Eliminación de años sin cobertura transaccional uniforme (`dropna`).

```python
# 1. Extraemos los ingresos de los datos brutos guardados
datos_ingresos = {}

for ticker in lista_saas:
    finanzas = datos_empresas[ticker]
    if 'Total Revenue' in finanzas.index:
        serie_ingresos = finanzas.loc['Total Revenue']
        # Nos quedamos únicamente con el AÑO (YYYY)
        serie_ingresos.index = pd.to_datetime(serie_ingresos.index).year
        datos_ingresos[ticker] = serie_ingresos

# 2. Consolidamos y convertimos a Billones ($B)
df_ingresos = pd.DataFrame(datos_ingresos)
df_ingresos_billions = (df_ingresos / 1e9).round(2).sort_index()

# 3. LIMPIEZA: Eliminamos las filas donde TODOS los datos sean NaN/N/A (como 2021)
df_ingresos_billions = df_ingresos_billions.dropna(how='all')

# 4. FORMATO PROFESIONAL PARA PRESENTACIÓN (.map para evitar warnings)
df_presentacion = df_ingresos_billions.map(lambda x: f"${x:.2f}B" if pd.notnull(x) else "N/A")

print("--- TABLA FINAL DE INGRESOS HISTÓRICOS (En Billones de USD / $B) ---")
df_presentacion
```

### 📐 Justificación Metodológica: Estandarización Transversal (Año Base 2025)

Al realizar un análisis comparativo (*Benchmark*) entre empresas con diferentes calendarios de reporte financiero, surge un problema de **asimetría temporal**:

1. **Desfase de Cierre Fiscal:** `CRM` reporta datos hasta el 31 de enero de 2026, mientras que `SHOP`, `TEAM` y `NOW` cierran sus libros al 31 de diciembre de 2025.
2. **Criterio de Estandarización (Manzanas con Manzanas):**
   * Si comparáramos el último dato disponible de cada una, le otorgaríamos a `CRM` 12 meses adicionales de acumulación de ingresos sobre el resto.
   * **Decisión:** Fijamos el **año 2025 como Periodo Base Estandarizado**, ya que representa el punto de corte homogéneo más reciente donde las 4 empresas cuentan con estados financieros auditados y completos de 12 meses.
3. **Tratamiento de Series Temporales:** Para la gráfica de líneas, la serie se mantiene completa hasta su último dato disponible, permitiendo observar la tendencia continua de cada firma.

##Financial_Modeling

###🔮 Proyección de Ingresos SaaS

---

### 🎯 Objetivo
Construir un modelo financiero proyectado para estimar los ingresos de las 4 empresas SaaS durante los próximos 3 años ($2026 - 2028$), evaluando su Tasa de Crecimiento Anual Compuesto ($CAGR$) y proyectando escenarios de escalamiento.

---

### 🐥 Explicación sencilla (En palabras simples)
Usaremos la historia de crecimiento de cada empresa (sus datos de 2022 a 2025) para calcular su "velocidad de crucero" ($CAGR$). Luego, le pediremos a Python que dibuje una línea hacia el futuro para ver cuánto dinero ganarán en 2026, 2027 y 2028.

---

### 📐 Formulación Matemática

1. **Tasa de Crecimiento Anual Compuesto ($CAGR$):**
   $$CAGR = \left( \frac{\text{Ingreso Final}}{\text{Ingreso Inicial}} \right)^{\frac{1}{n}} - 1$$
   *Donde $n$ es el número de periodos (años).*

2. **Fórmula de Proyección Futura:**
   $$\text{Ingreso}_{t+k} = \text{Ingreso}_{t} \times (1 + CAGR)^k$$

```python


# -----------------------------------------------------------------------------
# PASO 1: CÁLCULO DEL CAGR HISTÓRICO (2022 - 2025)
# -----------------------------------------------------------------------------
cagr_dict = {}
anios_hist = [2022, 2023, 2024, 2025]
n_periodos = len(anios_hist) - 1  # 3 años de diferencia

for ticker in df_ingresos_billions.columns:
    ingreso_inicial = df_ingresos_billions.loc[2022, ticker]
    ingreso_final = df_ingresos_billions.loc[2025, ticker]

    # Si falta el dato de 2022 (como en CRM), usamos 2023 como base ajustando periodos
    if pd.isna(ingreso_inicial):
        ingreso_inicial = df_ingresos_billions.loc[2023, ticker]
        n_adj = 2
    else:
        n_adj = n_periodos

    cagr = (ingreso_final / ingreso_inicial) ** (1 / n_adj) - 1
    cagr_dict[ticker] = cagr

# -----------------------------------------------------------------------------
# PASO 2: PROYECCIÓN A 3 AÑOS (2026, 2027, 2028)
# -----------------------------------------------------------------------------
anios_futuros = [2026, 2027, 2028]
df_proyeccion = df_ingresos_billions.copy()

for ticker in df_ingresos_billions.columns:
    base_2025 = df_ingresos_billions.loc[2025, ticker]
    cagr = cagr_dict[ticker]

    for i, anio in enumerate(anios_futuros, start=1):
        # Si la celda ya tiene dato real (como CRM 2026), proyectamos desde 2025 de forma consistente
        df_proyeccion.loc[anio, ticker] = round(base_2025 * ((1 + cagr) ** i), 2)

# -----------------------------------------------------------------------------
# PASO 3: VISUALIZACIÓN DE RESULTADOS & PROYECCIÓN
# -----------------------------------------------------------------------------
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

for ticker in df_proyeccion.columns:
    # Separar datos históricos vs proyectados
    hist = df_ingresos_billions[ticker].dropna()
    proy = df_proyeccion.loc[[2025, 2026, 2027, 2028], ticker]

    # Graficar histórico en línea sólida
    plt.plot(hist.index, hist.values, marker='o', linewidth=2.5, label=f"{ticker} (CAGR: {cagr_dict[ticker]*100:.1f}%)")
    # Graficar proyección en línea punteada
    plt.plot(proy.index, proy.values, linestyle='--', marker='s', linewidth=2, color=plt.gca().lines[-1].get_color())

plt.title("🔮 Modelo Predictivo de Ingresos SaaS (2026 - 2028)", fontsize=14, fontweight='bold', pad=15)
plt.ylabel("Ingresos Totales en Billones de USD ($B)", fontsize=12)
plt.xlabel("Año Fiscal", fontsize=12)
plt.legend(title="Empresa & CAGR", frameon=True)
plt.xticks([2022, 2023, 2024, 2025, 2026, 2027, 2028])
plt.show()

# Mostrar la tabla formateada con las proyecciones
print("--- TABLA DE PROYECCIONES FINANCIERAS (En Billones de USD / $B) ---")
df_proyeccion_formateada = df_proyeccion.map(lambda x: f"${x:.2f}B" if pd.notnull(x) else "N/A")
df_proyeccion_formateada
```

##Budgeting

### 💰Estructura de Gastos y Márgenes Operativos

---

### 🎯 Objetivo
Analizar la eficiencia operativa histórica de las empresas SaaS mediante la extracción de sus Gastos Operativos (*OpEx*) e Ingresos Operativos (*Operating Income*), proyectando un presupuesto estructurado para el periodo $2026 - 2028$.

---

### 🐥 Explicación sencilla (En palabras simples)
No basta con saber cuánto vende una empresa; hay que ver cuánto gasta para ganar ese dinero. Evaluaremos qué porcentaje de sus ingresos se va en mantener la empresa funcionando (*OpEx*) y cuánto le queda como ganancia limpia (*Margen Operativo*).

---

### 📐 Formulación Financiera

1. **Gastos Operativos ($OpEx$):** Suma de costos de investigación y desarrollo ($R\&D$), ventas y mercadeo ($S\&M$) y gastos administrativos ($G\&A$).
2. **Utilidad Operativa ($EBIT$):**
   $$\text{Utilidad Operativa} = \text{Ingresos Totales} - \text{Costos Directos} - OpEx$$
3. **Margen Operativo (\%):**
   $$\text{Margen Operativo} = \left( \frac{\text{Utilidad Operativa}}{\text{Ingresos Totales}} \right) \times 100$$

```python

plt.figure(figsize=(10, 5))
sns.set_theme(style="whitegrid")

# 1. Asignamos colores personalizados según el rendimiento financiero
# Verde para positivos, Rojo/Coral para la deficiencia de TEAM (márgenes negativos)
colores_custom = ['#2e7d32' if val > 0 else '#d32f2f' for val in df_budget_2025['Margen Operativo (%)']]

# 2. Creamos el gráfico de barras aplicando la paleta de colores alertas
ax = sns.barplot(
    x=df_budget_2025.index,
    y=df_budget_2025['Margen Operativo (%)'],
    palette=colores_custom
)

plt.title("📊 Eficiencia de Presupuesto: Margen Operativo (%) en 2025", fontsize=14, fontweight='bold', pad=15)
plt.ylabel("Margen Operativo (%)", fontsize=12)
plt.xlabel("Empresa (Ticker)", fontsize=12)

# Linea de corte horizontal en Y = 0 (Línea de equilibrio / Break-even)
plt.axhline(0, color='black', linewidth=1.2, linestyle='--')

# 3. Colocamos las etiquetas numéricas alineadas correctamente según la barra
for p in ax.patches:
    altura = p.get_height()
    # Si la barra es negativa (TEAM), ponemos el texto un poco más abajo para que no tape la barra
    offset = -15 if altura < 0 else 8

    ax.annotate(
        f"{altura:.1f}%",
        (p.get_x() + p.get_width() / 2., altura),
        ha='center', va='center',
        xytext=(0, offset), textcoords='offset points',
        fontweight='bold',
        color='black'
    )

plt.show()

print("--- TABLA RESUMEN DE PRESUPUESTO Y RENTABILIDAD (2025) ---")
df_budget_2025
```

##Forecasting

### 📈 Pronósticos Avanzados con Machine Learning

---

### 🎯 Objetivo
Desarrollar un modelo predictivo mediante Regresión Lineal (`LinearRegression` de `scikit-learn`) para realizar el pronóstico (*Forecasting*) de ingresos futuros ($2026 - 2028$) y evaluar la precisión del modelo utilizando el Coeficiente de Determinación ($R^2$).

---

### 🐥 Explicación sencilla (En palabras simples)
Entrenaremos a un algoritmo de Inteligencia Artificial / Machine Learning para que "vea" el comportamiento pasado de los ingresos y trace la curva de proyección estadística con mayor precisión técnica.

---

### 📐 Formulación Matemática del Modelo (Econometría, Micro y Estadística)
$$\hat{Y} = \beta_0 + \beta_1 \cdot X$$
* $\hat{Y}$: Ingreso proyectado en Billones de USD ($B$).
* $X$: Año fiscal.
* $\beta_1$: Pendiente de la recta (Velocidad de aceleración en $B$/año).
* $\beta_0$: Intercepto.
* $R^2$: Métrica que mide qué tan bien se ajustan los datos reales a la línea de regresión (donde $1.0$ es un ajuste perfecto).

```python

from sklearn.linear_model import LinearRegression

# -----------------------------------------------------------------------------
# PASO 1: ENTRENAR MODELO DE REGRESIÓN LINEAL POR CADA EMPRESA
# -----------------------------------------------------------------------------
anios_hist = np.array([2022, 2023, 2024, 2025]).reshape(-1, 1)
anios_futuros = np.array([2026, 2027, 2028]).reshape(-1, 1)
anios_totales = np.vstack((anios_hist, anios_futuros)).flatten()

df_forecasting = pd.DataFrame(index=anios_totales)
r2_scores = {}

for ticker in df_ingresos_billions.columns:
    # Extraemos serie limpia eliminando NaN (para CRM usamos 2023-2026 si aplica)
    serie_clean = df_ingresos_billions[ticker].dropna()
    X = np.array(serie_clean.index).reshape(-1, 1)
    y = serie_clean.values

    # Creamos y entrenamos el modelo de Machine Learning
    model = LinearRegression()
    model.fit(X, y)

    # Evaluamos la precisión con R^2
    r2_scores[ticker] = model.score(X, y)

    # Pronosticamos los valores para toda la línea de tiempo (2022 a 2028)
    predicciones = model.predict(np.array(anios_totales).reshape(-1, 1))
    df_forecasting[ticker] = np.round(predicciones, 2)

# -----------------------------------------------------------------------------
# PASO 2: VISUALIZACIÓN DEL FORECASTING CON BANDA DE TENDENCIA
# -----------------------------------------------------------------------------
plt.figure(figsize=(14, 6))
sns.set_theme(style="whitegrid")

for ticker in df_ingresos_billions.columns:
    # Puntos reales
    real = df_ingresos_billions[ticker].dropna()
    plt.scatter(real.index, real.values, s=60, zorder=5)

    # Línea de pronóstico estadístico (ML)
    plt.plot(
        df_forecasting.index,
        df_forecasting[ticker],
        linestyle='--',
        linewidth=2,
        label=f"{ticker} (ML Forecast | $R^2$: {r2_scores[ticker]:.2f})"
    )

plt.title("📈 Financial Forecasting: Regresión Lineal de Ingresos SaaS (2022 - 2028)", fontsize=14, fontweight='bold', pad=15)
plt.ylabel("Ingresos Totales en Billones de USD ($B)", fontsize=12)
plt.xlabel("Año Fiscal", fontsize=12)
plt.legend(title="Empresa & Métrica R²", frameon=True)
plt.xticks(anios_totales)
plt.show()

# -----------------------------------------------------------------------------
# PASO 3: TABLA DE PRONÓSTICOS ($B)
# -----------------------------------------------------------------------------
print("--- PRONÓSTICO DE INGRESOS CON MACHINE LEARNING ($B) ---")
df_forecast_presentacion = df_forecasting.map(lambda x: f"${x:.2f}B" if pd.notnull(x) else "N/A")
df_forecast_presentacion
```

##Scenario_Analysis

### 🎭Evaluación de Riesgo y Sensibilidad financiera

---

### 🎯 Objetivo
Evaluar la resiliencia y variabilidad de los ingresos proyectados para el año $2028$ mediante la simulación de 3 escenarios de mercado: **Base**, **Optimista (Bull)** y **Pesimista (Bear)**.

---

### 🐥 Explicación sencilla (En palabras simples)
Dado que el mercado tecnológico es volátil, no podemos confiar en una sola predicción. Simulamos qué pasaría con los ingresos en el año 2028 si las condiciones macroeconómicas y del mercado SaaS son favorables (+15%) o desfavorables (-15%).

---

### 📐 Definición de Escenarios para 2028

* **Escenario Base ($Base$):** Proyección estándar del modelo de Regresión Lineal/ML.
* **Escenario Optimista ($Bull\ Case$):**
  $$\text{Ingreso}_{\text{Bull}} = \text{Ingreso}_{\text{Base}} \times 1.15 \quad (+15\%)$$
* **Escenario Pesimista ($Bear\ Case$):**
  $$\text{Ingreso}_{\text{Bear}} = \text{Ingreso}_{\text{Base}} \times 0.85 \quad (-15\%)$$

```python

# -----------------------------------------------------------------------------
# PASO 1: CONSTRUCCIÓN DE LA TABLA DE ESCENARIOS AÑO 2028
# -----------------------------------------------------------------------------
# Tomamos la proyección base del modelo ML para el año 2028 (df_forecasting)
base_2028 = df_forecasting.loc[2028]

# Calculamos los 3 escenarios
bull_2028 = (base_2028 * 1.15).round(2)   # +15% Crecimiento acelerado
bear_2028 = (base_2028 * 0.85).round(2)   # -15% Contracción / Recesión

# Consolidamos en un DataFrame de Escenarios
df_escenarios_2028 = pd.DataFrame({
    'Bear Case (-15%)': bear_2028,
    'Base Case (ML)': base_2028,
    'Bull Case (+15%)': bull_2028
}).sort_values(by='Base Case (ML)', ascending=False)

# -----------------------------------------------------------------------------
# PASO 2: VISUALIZACIÓN DE RANGOS DE RIESGO (GRÁFICO DE BARRAS AGRUPADAS)
# -----------------------------------------------------------------------------
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")

# Convertimos la tabla a formato largo (melt) para seaborn
df_melted = df_escenarios_2028.reset_index().melt(
    id_vars='index',
    var_name='Escenario',
    value_name='Ingresos ($B)'
)
df_melted.rename(columns={'index': 'Empresa'}, inplace=True)

# Graficamos las tres barras por empresa
ax = sns.barplot(
    data=df_melted,
    x='Empresa',
    y='Ingresos ($B)',
    hue='Escenario',
    palette=['#e53935', '#1e88e5', '#43a047'] # Rojo (Bear), Azul (Base), Verde (Bull)
)

plt.title("🎭 Análisis de Escenarios de Ingresos para el Año 2028 ($B)", fontsize=14, fontweight='bold', pad=15)
plt.ylabel("Ingresos Proyectados ($B USD)", fontsize=12)
plt.xlabel("Empresa (Ticker)", fontsize=12)
plt.legend(title="Escenario Macro", frameon=True)

# Colocar valores sobre cada barra
for p in ax.patches:
    altura = p.get_height()
    if altura > 0:
        ax.annotate(
            f"${altura:.1f}B",
            (p.get_x() + p.get_width() / 2., altura),
            ha='center', va='center',
            xytext=(0, 7), textcoords='offset points',
            fontsize=9, fontweight='bold'
        )

plt.tight_layout()
plt.show()

# -----------------------------------------------------------------------------
# PASO 3: MOSTRAR TABLA DE ESCENARIOS FORMATEADA
# -----------------------------------------------------------------------------
print("--- RESUMEN DE SENSIBILIDAD Y ESCENARIOS AÑO 2028 ($B) ---")
df_escenarios_presentacion = df_escenarios_2028.map(lambda x: f"${x:.2f}B")
df_escenarios_presentacion
```
